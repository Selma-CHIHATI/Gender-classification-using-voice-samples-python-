# Data Preparation 

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, StandardScaler
import joblib  # Pour sauvegarder les objets de prétraitement

# Configuration pour la reproductibilité
np.random.seed(42)

# Chargement des données
dataset = pd.read_csv("/home/sarah/Desktop/gender-classifier-using-voice/PFE/PRECISION90.csv")
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values

# Gestion des données manquantes
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
X = imputer.fit_transform(X)

# Encodage des données catégorielles
encoder = LabelEncoder()
y = encoder.fit_transform(y)

# Normalisation des caractéristiques
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Sauvegarder les objets
joblib.dump(imputer, 'imputer.joblib') 
# Sauvegarder l'encodeur
joblib.dump(encoder, 'encoder.joblib') 
joblib.dump(scaler, 'scaler.joblib')

print("Data preparation and saving of preprocessing objects completed successfully.")
```

```python
import numpy as np
import time
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
import matplotlib.pyplot as plt
import seaborn as sns

# Creating an instance of the SVM model

classifier = SVC(kernel='rbf', C=10.33229560435906, gamma='scale', shrinking=True, tol=0.018084807463876782, class_weight=None, random_state=42)

# Start timing
start_time = time.time()

# Obtaining predictions using cross-validation
y_pred = cross_val_predict(classifier, X, y, cv=4, n_jobs=-1)

# End timing
execution_time = time.time() - start_time

# Generating the confusion matrix
cm = confusion_matrix(y, y_pred)
plt.figure(figsize=(5, 3))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix - SVM with Gaussian Kernel')
plt.xlabel('Predicted Values')
plt.ylabel('True Values')
plt.show()

# Calculating accuracy
accuracy = accuracy_score(y, y_pred)
print("Accuracy: {:.2f}%".format(accuracy * 100))

# Calculating precision, recall, and F1-score specifically for classes 0 and 1
precision_0 = precision_score(y, y_pred, pos_label=0)
precision_1 = precision_score(y, y_pred, pos_label=1)
recall_0 = recall_score(y, y_pred, pos_label=0)
recall_1 = recall_score(y, y_pred, pos_label=1)
f1_0 = f1_score(y, y_pred, pos_label=0)
f1_1 = f1_score(y, y_pred, pos_label=1)

# Calculating average metrics
avg_precision = (precision_0 + precision_1) / 2
avg_recall = (recall_0 + recall_1) / 2
avg_f1 = (f1_0 + f1_1) / 2



# Printing the metrics
print(f"Precision for class 0: {precision_0}, Precision for class 1: {precision_1}")
print(f"Recall for class 0: {recall_0}, Recall for class 1: {recall_1}")
print(f"F1-score for class 0: {f1_0}, F1-score for class 1: {f1_1}")
print(f"Average Precision: {avg_precision}")
print(f"Average Recall: {avg_recall}")
print(f"Average F1-Score: {avg_f1}")
print(f"Execution Time: {execution_time} seconds")
```

```python
import numpy as np
import optuna
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

def objective(trial):
    # Définition de la plage des valeurs pour chaque paramètre
    n_estimators = trial.suggest_int('n_estimators', 10, 300)
    max_depth = trial.suggest_int('max_depth', 1, 50)
    min_samples_split = trial.suggest_int('min_samples_split', 2, 20)
    min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 20)
    max_features = trial.suggest_categorical('max_features', ['sqrt', 'log2', None])
    max_leaf_nodes = trial.suggest_int('max_leaf_nodes', 10, 1000, log=True)
    bootstrap = trial.suggest_categorical('bootstrap', [True, False])

    classifier = RandomForestClassifier(
        n_estimators=n_estimators,
        max_depth=max_depth,
        min_samples_split=min_samples_split,
        min_samples_leaf=min_samples_leaf,
        max_features=max_features,
        max_leaf_nodes=max_leaf_nodes,
        bootstrap=bootstrap,
        random_state=42,  # For reproducibility
        n_jobs=-1  # Use all available cores
    )

    # Évaluation du classificateur par validation croisée
    score = cross_val_score(classifier, X, y, cv=3, scoring='accuracy', n_jobs=-1)
    return np.mean(score)

# Chargement des données préparées (supposez que vous avez déjà X et y)
# X, y = votre_fonction_de_chargement_des_données()

# Création d'une étude Optuna pour trouver les meilleurs hyperparamètres
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=100)
print("Meilleurs paramètres:", study.best_trial.params)
print("Meilleure précision de validation croisée: {:.2f}%".format(study.best_value * 100))
```
